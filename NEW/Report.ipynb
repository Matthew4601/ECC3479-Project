# Report

The 2022 inflation surge was the largest across advanced economies in forty years, forcing central banks to raise interest rates at the fastest pace since the 1980s. A central question for policymakers was whether the shock represented a temporary energy price spike that would self-resolve once energy markets stabilised, or whether it had already fed into the broader price level through second-round effects — wage growth, services prices, and core consumer prices. That distinction determined both how aggressively to tighten monetary policy and for how long. The divergence in outcomes across the United States, United Kingdom, and Canada sharpened this question further: all three economies were exposed to the same global energy price shock, yet their core inflation paths differed in timing, magnitude, and persistence. This study investigates whether those divergences reflect structural differences in how energy prices transmit into core inflation across the three countries, and whether the 2022 episode represented a genuine departure from the preceding thirty-five years of suppressed energy pass-through.

Energy pass-through is the degree to which a rise in energy prices feeds into the broader price level — specifically core CPI, which excludes food and energy from its basket. A high pass-through means that when energy becomes expensive, firms raise prices across other goods and services to cover higher production and transport costs, broadening inflation beyond the energy sector. A near-zero pass-through means energy price movements remain largely contained.

Core inflation pass-through from energy prices was near-zero across all three countries from the mid-1980s through to 2020. This followed a period of large positive pass-through during the 1970s oil shocks, when energy price surges fed directly into broader inflation across all three economies. That transmission link then collapsed through the Great Moderation as central banks established greater credibility, production became less energy-intensive, and inflation expectations became better anchored. Pass-through remained suppressed through the 2000s and 2010s.

The 2022 shock broke from this pattern for Canada and the United Kingdom, but not convincingly for the United States. In Canada, the energy-to-core CPI pass-through coefficient shifted from −0.066 before 2022 to +0.065 after — a sign reversal indicating that energy price rises began to co-move positively with core inflation from 2022 onwards. The United Kingdom showed an almost identical shift, from −0.097 to +0.031. Both results survive a range of robustness checks, including autocorrelation-corrected standard errors, alternative break date specifications, exclusion of the COVID period, and sub-period analysis, making them among the most reliable findings in the study.

The United States did not exhibit the same change. The estimated shift in pass-through was small and negative (−0.040), meaning the pre-2022 positive pass-through of +0.098 fell modestly to +0.058 after 2022 rather than reversing direction. This result does not survive correction for autocorrelation, nor does it hold across alternative break date specifications, indicating that no robust structural change in the transmission mechanism can be identified for the United States over this period.

The 2022 episode was also historically moderate in scale. An episode comparison regression finds that the extra transmission recorded in 2022 (+0.024) was substantially smaller than during the 1973 oil shock (+0.088) and the 1979 shock (+0.054). Better-anchored inflation expectations and prompt monetary tightening appear to have limited the second-round effects that made the 1970s so damaging.

Once the persistence of core inflation itself is controlled for, the cumulative dynamic pass-through is economically negligible across all three countries. Core inflation momentum — where inflation stood last month — accounts for nearly all short-run variation. Outside of specific shock episodes, energy prices do not persistently drive core inflation in any of the three countries. The 2022 result therefore reflects a change in contemporaneous co-movement during a discrete shock window, not evidence of a sustained energy-driven inflation dynamic.

Taken together, the findings indicate that the 2022 inflation shock was structurally distinct from the three preceding decades of relative stability, but did not replicate the breadth or persistence of the 1970s episodes. Canada and the United Kingdom, with greater domestic exposure to wholesale energy prices, experienced genuine shifts in how energy costs transmitted into core inflation. The United States, where domestic energy markets are more insulated from global price movements and monetary policy responded swiftly, did not. This divergence is consistent with the energy system composition of each country playing a role in determining pass-through, though the panel interaction between pass-through and fossil fuel share cannot formally confirm this mechanism with only three countries in the sample.
