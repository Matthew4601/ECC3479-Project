Was the 2022 inflation shock fundamentally different from prior episodes, and to what extent did divergences in core inflation trends across the UK, US and Canada reflect differences in each country’s energy generation systems and their vulnerability to energy price pass-through?

The 2022 inflation surge was the most severe period of price instability across advanced economies since the 1980s. This is sharpened by a cross-country analysis: the United Kingdom, Canada and the United States were each exposed to the same global energy price surge following Russia’s 2022 invasion of Ukraine, yet their core inflation trajectories differed in timing, peak magnitude and the speed of the subsequent return to normalcy. Core inflation, which excludes food and energy from its makeup, is therefore the standard measure of whether energy shocks have fed through into the broader price level, capturing differences in peak timing and responses to other factors both exogenous and endogenous within each nation. If the shock were purely global and symmetric then core inflation rates should have responded similarly; the fact that they did not implies that domestic structural factors mediated the transmission of the global shock in country-specific ways.
This analysis investigates whether cross-country divergences reflect systematic differences in energy price pass-through — the degree to which movements in energy prices feed into core inflation — and whether the 2022 shock represented a genuine structural departure from the preceding thirty-five years of suppressed pass-through. The analysis uses monthly data from January 1971 to March 2025 and employs six regression specifications: a baseline OLS pass-through model, an autoregressive distributed lag model, a structural break specification, a panel model with country fixed effects and a fossil fuel share interaction, a rolling window estimation, and an episodic comparison regression that benchmarks 2022 against the shocks of 1973 and 1979. The approach is descriptive rather than causal, but the specifications and robustness checks provide a rigorous empirical characterisation of how the energy-to-core inflation relationship has evolved. The central finding is that pass-through, suppressed to near zero across all three countries from the mid-1980s through to 2020, increased in 2022 for Canada and the United Kingdom but not for the United States. In Canada and the United Kingdom the pass-through coefficient reverses sign after the 2022 shock, with both results surviving a range of robustness checks. The US, however, exhibits no robust structural change. The 2022 episode is then compared historically and found to be only around a quarter as intense as the shocks of the 1970s. The cross-country divergence is consistent with differences in domestic energy markets and monetary policy responses, though three countries are insufficient to formally confirm the precise structural mechanism through which these factors operate.



Data
The data was primarily sourced from the Federal Reserve Economic Data (FRED) and the OECD Main Economic Indicators (MEI) datasets. The core CPI levels were sourced through the FRED with the data for the United States needing to be converted from an index level into a year-on-year percentage change to ensure it was comparable with the Canadian and British datasets. Canadian and British energy CPI datasets were sourced from the MEI website, while the American energy CPI data was available from the FRED, although the datasets did not require any transformation to be placed on the same scaling. Additional datasets such as the variables used in later regressions (Brent crude oil price, Henry Hub gas spot prices and the electricity generation mix) were sourced from the International Energy Agency (IEA) and Our World in Data respectively. The three core cpi datasets were combined into a single dataset covering January 1971 to March 2025, with all three having these dates common. The energy cpi data was similarly restricted to this timeframe when combining the datasets to ensure comparability. The final dataset contains 651 monthly observations, with no data being dropped due to quality concerns or missing value issues. An issue to note is the interpretation of the coefficients produced from this data with the United Kingdom and Canada reporting as index levels with 2015 as the common base, whereas the United States reports as a YoY growth rate. Summary statistics are provided under the summary section of the appendices. 

Summary statistics
The United Kingdom core CPI data shows higher mean and variance compared to the United States and Canada, reflecting the larger inflation episodes of the 1970s and 1980s when the UK’s core CPI exceeded 20%. All three of the datasets are right skewed with the upper tail from the 1970s and 2022 episodes pulling the mean above the median statistic in each case.
The United Kingdom and Canada’s energy CPI series reflects the low energy price environment of the intervening period between the 1970s shocks and the 2022 shock. The large standard deviations indicate that these shocks dominate the long-run distribution. The effects of other non-energy crises are similarly viewable, most notably with the 28% contraction of the United States economy during the 2007–2008 global financial crisis.
The analysis is descriptive and estimates the conditional correlation between energy prices and core inflation, documenting how that correlation changes with the 2022 energy shock and comparing the change between the historical periods of the three examined nations. There is no established causal mechanism within this research as there is no credible source of exogenous variation to exploit, owing to confounding factors:, both core CPI and energy CPI for Canada and the UK are non-stationary, which would estimate long-term co-movement rather than a causal effect, and the residuals are strongly correlated owing to the common global shock that they all faced. Therefore this analysis makes the descriptive claim that the pattern of co-movement between energy prices and core inflation changed structurally during 2022 for the United Kingdom and Canada, and that this finding is robust across a range of alternative specifications.

Empirical Strategy
Regression specifications
Regression 1 Baseline pass-through
An OLS regression of core CPI on the energy CPI for each country separately, estimated over the full 1971–2025 sample period, with HC3 heteroskedasticity-robust standard errors. This regression establishes the long-run co-movement and provides the benchmark coefficient against which the other specifications are compared. 

Regression 2 Autoregressive Distributed Lag Model (ADL)
To capture dynamic and delayed pass-through, the original model is extended to include 12 lags of the energy cpi and one lag of the core CPI as an autoregressive term. The lagged core CPI term absorbs the strong persistence in core inflation, with the cumulative sum of the twelve lag coefficients measuring the medium-term pass-through once inflation momentum is controlled for. The regression directly tests whether energy prices carry information about the future path of core inflation beyond what past levels of core CPI predict.

Regression 3 Structural Break
A binary dummy (D2022) is added into this specification which equals one from January 2022 onwards, interacting with the first regression. The interaction coefficient (delta) captures whether the pass-through relationship changed after the 2022 shock, with the break date of January 2022 reflecting the point at which core CPI began to rise sharply in all three countries and is validated by the robustness tests.

Regression 4 Panel Fixed Effects With Fossil Fuel Share Interaction
All three countries are pooled into a single panel, with country fixed effects absorbing structural differences in average inflation levels across them. Model A estimates a common pass-through coefficient within the framework of the three controlled-for nations. Model B adds an interaction between the energy CPi and each country’s fossil fuel share of electricity generation, testing whether a greater reliance on fossil fuels amplified the transmission of energy CPI into core CPI. This test, specifically Model B, suffers from a limited data pool, containing only the three countries used in the rest of the study, which is insufficient to identify country-level structural differences in their respective energy makeup.

Regression 5 Rolling Window OLS
The original first regression is re-estimated in successive 60-month rolling windows, producing a time series for the beta estimates for each country between the years 1976 and 2025. This imposes no structural break date and provides the clearest picture of how the pass-through relationship evolved over time. This is used to contextualise the structural break findings from the third regression over a broader 50 year period.

Regression 6 Episode Comparison
Three additional binary dummies are added which control for the identified shocks in the 70’s and in 2022, with each interacting with the energy CPI and included within a pooled panel model with fixed country effects. This measures the pass-through coefficients during each historical shock relative to the baseline fixed effect. This determines whether the 2022 transmission was larger or smaller than its counterparts in the 1970’s, and whether the differences are statistically distinguishable from one another.

Results
Regression 1 results
The Canadian beta coefficient of −0.049 demonstrates that, when holding all else constant (ceteris paribus) a one unit rise in the energy price index is associated with a 0.049 percentage point (pp) fall in core CPI growth. This result is counterintuitive to what might have been expected but is rather indicative of an issue with the format of the dataset where the OECD energy index drifts upwards over time, whilst the FRED CPI growth is a stationary rate that tends downwards from the 1970s highs during the period between the 1970s and 2022 shocks. This issue is also encountered with the United Kingdom, as the data formatting is the same between the two nations with the beta coefficient for the UK being −0.057 pp.
The United States however, does not have the same issue as the previous two countries, with a beta of 0.094 meaning, ceteris paribus, a one percentage point rise in energy inflation is associated with a 0.094 pp rise in core inflation. This adheres to the expectation that energy costs would feed into production over time and drive up the core inflation rate.

Regression 3 Structural Break coefficients
Canada resulted in a delta of 0.131, meaning that from January 2022 onwards the pass-through coefficient increased by 0.131 units relative to the pre 2022 level. The pre-2022 coefficient of −0.066 coupled with this delta results in a post-2022 pass-through effect of +0.065 units (−0.066 + 0.131 = 0.065). The sign of the coefficient reverses, indicating that before 2022 higher energy price index levels were negatively correlated with core inflation growth over the full sample period, whereas after 2022 a higher energy price index level is associated with a higher core CPI growth rate of 0.065 pp per index unit. The United Kingdom has a similar result to Canada, with a delta of 0.128 resulting in a post-2022 pass-through effect of 0.031pp per index unit (-0.097 +  0.128 =0.031).
The United States, however, moved conversely to the United Kingdom and Canada, with a negative delta of −0.040, meaning that the pre-2022 pass-through fell slightly to 0.058 pp per index unit after 2022 (0.098 − 0.040 = 0.058). This result does not withstand autocorrelation corrected standard errors, unlike the United Kingdom and Canada’s results, and therefore no reliable conclusion for the United States can be drawn from this regression, even though all the coefficients were nominally significant.

Regression 6 Y episode coefficients
The Y1973 term shows that during the 1973–75 oil shock the energy-to-core pass-through was 0.088 units higher per period than the non-episodic baseline of 0.043. The Y1979 term, which covers the 1979–1982 shock, produced a coefficient 0.054 units above the non-episodic baseline and the Y2022 term had a coefficient of 0.024. This shows that out of the three energy shocks within the dataset, the 2022 shock had the least intense impact in pushing the countries above the baseline pass-through, roughly a quarter of the intensity of the 1973-75 shock. This still demonstrates that the 2022 shock had an impact in raising pass-through just not to the extent of the previous shocks.

Economic Significance
The Canadian energy price index rose from approximately 60 in early 2021 to a peak of near 110 in mid-2022, a rise of 50 index units. Applying the post-2022 pass-through of 0.065 per index unit, 3.25 pp (50 × 0.065) of core CPI is attributed to the energy price surge. Canadian core inflation peaked at 6.2% in mid-2022 against the pre-shock baseline of around 2% (central bank target), leading to an overshoot of 4.2pp (6.2-2). The energy pass-through calculation accounts for around three quarters of the overshoot above the inflation target. For the United Kingdom the energy index rose by 130 units, factoring in the post-2022 pass-through this leads to 4.03pp, which accounts for the majority of the UK’s overshoot of their 2% inflation target.
The ADL model shows a sum of betas that is equivalent to -0.001 for Canada and the the UK, with a positive 0.003 for the United States when controlling for lagged core inflation. These effects are economically negligible, showing that under normal (non-shock) conditions a sustained 10 unit rise in the energy index constant for twelve months would move core cpi by 0.001pp for Canada and the UK, 0.003 for the US. This shows that the 2022 energy shock didn’t trigger a second-round dynamic, in which the higher energy shock fed into prices and other factors, keeping pass-through persistent and elevated. This is similarly reflected in the retreat of core inflation once energy prices began to fall after the initial shock period, the 2022 shock did not create a price spiral inducing persistently higher inflation.
The difference between the 2022 and 1973–75 shocks is non-trivial, with the 2022 pass-through being roughly a quarter of that seen in the 1970s. This has knock-on effects for the rest of the economy, if the magnitude of the impact had been significantly higher, institutions such as the Bank of England may have had to deploy more aggressive monetary policy actions than were already taken to lessen the long term burden.

Robustness
The results for Canada and the United Kingdom were generally robust. Delta terms were positive and significant in six of the seven robustness checks run to verify these coefficients. They are both insensitive to the exact break date, surviving autocorrelation correction and the results hold after partitioning out the influence of COVID-19. The single test that they both failed is first difference which reflects whether months to month energy price changes predict month to month changes in core inflation.
The United States delta coefficients are not robust, they fail the HAC correction and alternative date specifications, and the sign reverses in the first differences test. Therefore no reliable conclusions can be drawn from the United States dataset.
Once time fixed effects are added to absorb shocks common to all three countries simultaneously (robustness test 10) the pooled beta coefficient loses significance (p = 0.345). This shows that part of the panel correlation is driven by global factors and not country specific transmission.
All three countries show positive and significant betas in the post-COVID era, with the near-zero pre-COVID result confirming that this represents a structural shift rather than just a long-run constant.
The ordering of shock magnitudes from the 1970s, 1980s, and 2022 survives time-trend control and sub-period analysis, and is consistent with the rolling betas in the fifth regression, where the post-shock means are below the 1970s peaks.

Discussion
The results present a clear and consistent picture for Canada and the United Kingdom, with energy-to-core CPI pass-through suppressed for roughly three and a half decades between the 1970s shock and the 2022 shock. The structural break coefficients are large, precisely estimated, and survive every robustness check that targets the United Kingdom and Canada’s results. The rolling window analysis shows that the re-elevation was not a spike but a sustained shift to a higher level, coupled with the episode comparison, confirm that the shift is real compared to the baseline, even if it is smaller in magnitude than the shocks in the 1970s. The United States, however, does not conform to the results shown by the United Kingdom and Canada with the absence of robust findings making it impossible to state any reliable relationship within the American results. The core finding that cumulative dynamic pass-through is near zero once core inflation momentum is controlled for means the 2022 shock produced a contemporaneous shift in core inflation rather than a sustained propagation feedback loop.

Limitations
The unit inconsistency across the energy series is the main limitation, with the UK and Canada energy CPI levels being index-based whilst the United States series is a YoY growth rate. This creates issues when comparing the betas as they’re not directly comparable in magnitude across countries and the pooled panel regressions combine non-similar units, which taints the result, as shown by the negative pooled 2022 pass-through, which is distorted by the index-level dominance. The workaround was using the change in pass through within each country rather than cross country comparison of beta levels, but this still limits the usefulness of the pooled specification
The analysis is descriptive and not causal; there is no instrument for energy prices, as the unobserved correlation between energy prices and core inflation could reflect reverse causality, or omitted variable bias from other factors in the supply chain. The structural break identifies co-movement, not a causal mechanism, meaning the findings are useful as evidence of conditional correlation rather than transmission.
The small number of countries used restricts the comparisons available to identify structural mechanisms within each country, as evidenced by the weak linking of fossil fuel share in the fourth regression.

Conclusion
For Canada and the United Kingdom the pass-through coefficient sign changed in 2022 from negative to positive with this shift being large, precise and robust. Applying the post-break pass-through to the actual energy price movements in 2022 accounts for the majority of the core inflation overshoot above target in both nations. The 2022 shock was therefore not simply a large energy price spike that left core inflation untouched but rather actively transmitted into the broader price level in a way that other price shocks, such as those in 2008, did not.
For the United States, no robust structural break can be identified; the result is fragile to standard error correction and inconsistent across alternative specifications, with no clear change in levels. This suggests that there are other US-only factors that influenced this divergence from the United Kingdom and Canada and mediated the 2022 shock for the US. What cannot be concluded is that the energy system composition caused divergence, with the fossil fuel panel too underpowered, containing only three countries to compare. The analysis also only documents that co-movement changed and does not establish why it changed.
Solutions to the limitations would include an expanded dataset with more countries to compare in terms of energy generation makeup. Instrumental analysis could also aid in assisting for a determination on a causal relationship between the energy CPI movements and core CPI changes.
