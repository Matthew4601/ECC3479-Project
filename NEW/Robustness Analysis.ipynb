This analysis is descriptive, not causal. The analysis provides rigorous descriptive evidence that energy prices and core CPI co-moved more strongly during the 2022 episode than during the prior low-inflation decade for Canada and the UK, that this pattern is robust across specification choices, and that country-level energy series capture variation beyond the global oil price. These findings motivate causal hypotheses about transmission mechanisms and generation mix vulnerability but do not themselves identify causal effects.

Robustness Test 1.
Testing for serial correlation.
The regressions were rerun with HAC standard errors instead of the HC3 errors present in the initial regressions to see if serial correlation is present and whether anything stated as initially significant is in fact not.
After performing this test it shows that all three countries' pass-through coefficients remain significant at the 1% level. The structural break, however, only remains significant for the United Kingdom and Canada; the United States structural shift was not robust to autocorrelation.

Robustness Test 2.
Testing for spurious regressions and whether the data is stationary.
This checks the regressions for a spurious regression by testing whether the means are fixed (stationary).
Five of the six series are non-stationary, with only the USA Energy variable being stationary. This shows that the coefficients in these regressions could potentially be spurious and should be interpreted as a long-run co-trending relationship rather than a short-run causal one.

Robustness Test 3
Alternative Structural Break
Uses a different date for the structural break to test whether the direction and significance of the delta term is consistent for each nation. Canada and the United Kingdom are both robust and significant at the 1% level, regardless of the date used (which ranged between March 2020 and July 2022). The United States is not significant across any of the four dates. This shows that for the United Kingdom and Canada this is a broad phenomenon and not a date-specific result, though the same cannot be said for the United States.

Robustness Test 4
Endogenous Break Detection.
Removes any researcher input into determining dates that may have resulted in significant energy shocks in the economy. The data is then searched monthly for a date that produces the lowest combined residual sum of squares across the sub-periods. Using this, other breaks were identified within the data.
This shows that the 2022 shock is not the most statistically significant shock in the data, although it is still notable.

Robustness Test 5
COVID Exclusion.
COVID provided supply chain disruptions that influenced core inflation in ways not correlated with energy price transmission, and therefore there is a risk that this is represented in the 2022 data, distorting the results. Two tests were then run: the first includes a dummy variable representing January 2022 to December 2023 only; the second utilises a separate COVID dummy variable to partition the two time periods explicitly.
Canada and the United Kingdom show a strongly significant result in the pure 2022 window with no COVID effects. The United Kingdom has a separate COVID-response coefficient, distinct from the 2022 energy shock, that is large and significant. The United States is significant at the 5% level only in the second test where COVID is partitioned out. This shows that the United States break coefficient has some contamination from the COVID period and is not entirely attributable to the 2022 price shock.

Robustness Test 6.
Gas Price Check (US Only)
Given that the 2022 price shock derived primarily from increases in LNG prices, this test checks whether using the Henry Hub gas price (YoY%) as the energy measure for the USA, rather than the broad energy CPI, produces a materially different result. Neither the broad energy CPI nor the Henry Hub gas price is statistically significant for the USA in this subsample, and the Henry Hub measure produces a lower R² (0.015 vs 0.073). The structural break specification with Henry Hub is also not significant (p=0.671). This suggests that the USA pass-through result in Regression 1 is heavily dependent on the full sample period — particularly the 1970s — and that gas-specific prices do not better explain US core CPI in the more recent period.

Robustness Test 7
First-Difference Specification
Reruns the regressions using the month-on-month changes in each series rather than purely the level of core CPI on energy CPI, thus removing the trending component and making the series stationary (addressing the issues raised in Test 2), so that any relationship identified is a genuine short-run interaction rather than two variables that simply share a common trend.
The structural break is no longer present for Canada and the United Kingdom. Using this in unison with the results of Robustness Test 2, it shows that the results capture a long-run co-trending relationship, with core and energy CPI moving together over time, and that the 2022 structural shift is a persistent level break rather than a change in short-run month-to-month dynamics. The United States structural break is robust and significant at the 1% level, but is negative. This shows that in the United States the short-run dynamic itself changed in 2022 — the episode actually dampened month-to-month energy price transmission into core CPI, consistent with the Federal Reserve's aggressive rate response.

Robustness Test 8
Alternative Lag Lengths
In the second regression there are 12 lags for energy CPI to allow a reasonable amount of time for energy prices to transmit into core CPI. This test changes the number of lags from 12 to 6 and 24, providing comparison to a six-month and two-year period.
All three countries have a cumulative pass-through that is effectively zero across all lag lengths, showing that once core inflation momentum is accounted for, Regression 2 is fully robust.

Robustness Test 9
Controlling for the Brent Crude Oil Price.
Adding a control for the Brent Crude Oil price to determine whether the changes in energy CPI are just a proxy for changes in the Brent Crude Oil price and are not capturing country-specific transmission mechanisms. A significant Brent Crude Oil result would essentially state that the only factor in the energy CPI is the price of oil and not the energy generation methods of each country.
The Brent Crude variable is not significant for any of the countries, thus showing that changes in each nation's energy markets are not entirely driven by the change in the price of oil.

Robustness Test 10
Two-Way Fixed Effects
Tests whether the residuals are correlated across countries after model estimation in the fourth regression, where all three countries are pooled together. If the errors tend to be similar amongst the countries (i.e. negative or positive at the same time), this indicates a common shock they are simultaneously exposed to that is not captured in the model. With all three countries exposed to the 2022 shock, this would be expected, and would confirm that the 2022 energy shock affects all three countries.
There is strong cross-sectional dependence that is significant, indicating that the 2022 shock is definitively affecting all three nations, although this does not reveal the degree to which it affects each country individually.

Robustness Test 11.
Sub-Period Stability and Time-Trend Control.
Directly probes whether the full-sample conditional correlation in Regression 1 is driven by a particular historical epoch or functional-form choice — the key credibility check for a descriptive claim. Part A re-estimates β separately across four historical epochs (oil shocks 1971–84, Great Moderation 1985–99, pre-COVID 2000–19, post-COVID 2020–25) under HAC standard errors. Part B adds a linear time trend to Regression 1 to test whether the correlation survives once a shared deterministic trend is removed.

## Robustness Interpretation

---

### Cluster 1 — Autocorrelation, Non-Stationarity, and First Differences (Tests 1, 2, 7)

The baseline energy pass-through β survived HAC correction at the 1% level for all three countries, confirming the levels result in Regression 1 is not an artefact of serial autocorrelation. The structural break δ also survived HAC correction for Canada and the UK, but the USA break lost significance entirely under HAC (p=0.250), exposing it as dependent on the choice of HC3 errors. In first differences, the break disappears for Canada and the UK while becoming significant but negative for the USA.

The disappearance of the structural break in first differences for Canada and the UK is not a failure — it confirms that the 2022 episode produced a persistent level shift in the long-run pass-through relationship rather than a change in short-run month-to-month dynamics, which is the stronger and more economically meaningful result. The USA pattern is the inverse: no long-run break, but a significant negative short-run dynamic break, consistent with the Federal Reserve's aggressive rate response dampening the month-to-month transmission channel. The key credibility problem this cluster creates is for the USA: its Regression 3 result should not be treated as robust and must be heavily qualified in any interpretation.

---

### Cluster 2 — Break Date Sensitivity and COVID Contamination (Tests 3, 4, 5)

Canada and the UK structural break δ is significant at the 1% level under every alternative break date (March 2020, July 2021, January 2022), in the strict 2022-only window, and with COVID partitioned out separately. The USA is not significant under any alternative break date, and only reaches 5% significance when COVID is explicitly separated. Test 4's endogenous detection identifies the largest statistical break in the full sample at 1983 (Canada) and 1992 (UK and USA) — not 2022 — and the PELT algorithm detects a break at March 2020 for all three countries rather than January 2022 specifically.

For Canada and the UK this cluster provides strong independent validation: the result is not an artefact of a convenient date choice, is not driven by COVID-era supply disruptions, and is stable in sign and magnitude across six different specifications. Test 4 contextualises the broader narrative: the 2022 shock is not the dominant structural break in the historical record — that distinction belongs to the exit from the 1970s inflation regime. The 2022 episode is better described as a partial re-activation of a transmission channel dormant since the early 1980s. The PELT detection of March 2020 rather than January 2022 further suggests the regime shift began with COVID disruptions and the 2022 energy shock extended it, which is consistent with the break being robust whether the dummy starts at 2020 or 2022.

---

### Cluster 3 — Alternative Measures and Global Factors (Tests 6, 9)

Test 9 shows that Brent crude oil is not significant for any country once the domestic energy CPI is included, and the country-level β is essentially unchanged when Brent is added. Test 6 shows that in the post-1997 subsample where Henry Hub data is available, neither the broad energy CPI nor the gas-specific price is significant for the USA, and the structural break with Henry Hub is also insignificant (p=0.671).

Test 9 directly supports the dissertation's comparative argument: the country energy CPI is capturing something genuinely country-specific — reflecting generation mix, import dependence, and domestic pricing arrangements — rather than simply re-expressing the global oil price. This strengthens the conceptual basis for the cross-country divergence narrative. Test 6 is the most uncomfortable result in the analysis: the USA's strong positive β in Regression 1 (+0.094) is essentially a 1970s phenomenon that disappears in the more recent period regardless of the energy measure used. The USA therefore cannot be presented as a symmetric third case — it functions as a contrast rather than a parallel, with a structurally different post-1997 energy-core relationship that likely reflects more domestically insulated energy markets and better-anchored inflation expectations.

---

### Cluster 4 — Dynamic Pass-Through Lag Robustness (Test 8)

Cumulative pass-through is near zero for all three countries at 6, 12, and 24 lags. No lag length produces a meaningful estimate.

This is the cleanest result in the robustness analysis. The choice of 12 lags was not driving Regression 2's finding, and the conclusion that energy prices have no meaningful cumulative dynamic pass-through to core inflation — once inflation's own persistence is controlled — holds at every horizon tested. The implication is that the connection between energy prices and core inflation operates at the level of long-run regimes and structural conditions rather than through a simple lagged transmission process, with the AR(1) coefficient absorbing virtually all of the inflation dynamics.

---

### Cluster 5 — Cross-Sectional Dependence (Test 10)

The Pesaran CD statistic (29.39, p=0.000) and pairwise residual correlations of 0.74–0.79 confirm strong common-shock structure across all three countries. Under two-way fixed effects, which absorbs both country and time-period variation, the panel energy coefficient becomes insignificant (p=0.345).

The loss of significance under two-way FE is expected rather than damaging: adding time fixed effects in a three-country panel extracts the common time-series variation that all three countries share, which in this context is the 2022 shock itself — the variation being studied. The more meaningful result is the cross-sectional dependence finding, which confirms all three economies were simultaneously exposed to a large common shock. The fact that Canada and the UK show substantially larger structural breaks than the USA in response to that common shock therefore reflects country-specific differences in transmission rather than differences in shock exposure. This reinforces the country-level regressions as the primary identification strategy, with the panel regressions serving a secondary comparative role.

---

### Cluster 6 — Sub-Period Stability and Functional Form (Test 11)

The full-sample β varies dramatically in sign and significance across epochs: Canada and the UK's negative full-sample coefficient is dominated by the Great Moderation period (β ≈ −0.26 to −0.31), while the USA's positive full-sample β is driven almost entirely by the oil-shock era (β = +0.175). All three converge to a positive, significant β post-COVID (0.053–0.065). After adding a linear time trend, Canada and the UK's β flips positive (+0.082, +0.131), revealing that their negative full-sample coefficient is partly a trend artefact from pooling the high-inflation 1970s with the low-inflation 2000s; significance is unchanged for all three countries.

This is the check most directly relevant to the credibility of the descriptive claim. The finding that the full-sample β is subsample-dependent does not undermine the analysis — it sharpens it. The research question is about the 2022 episode specifically, and the post-COVID epoch (the most relevant window) shows a consistent, positive, and significant pass-through for all three countries with no sign ambiguity. The Great Moderation period's negative β reflects mean-reversion dynamics in the long-run trend and is not the relationship being claimed. The time-trend result confirms the correlation survives functional-form variation and is not a shared deterministic trend artefact. Together, Test 11 supports presenting the results epoch-specifically rather than relying on full-sample pooled coefficients, which the negative signs in Regression 1 for Canada and the UK would otherwise make difficult to interpret.
